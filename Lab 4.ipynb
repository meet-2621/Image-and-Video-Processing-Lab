1. WAP to enhance the input image using Log Transform Assume c = 1.       

```python
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg")
c = 1
log_img = c * np.log1p(img.astype(np.float32))
log_img = cv2.normalize(log_img, None, 0, 255, cv2.NORM_MINMAX)
log_img = np.uint8(log_img)
cv2.imshow("Original", img)
cv2.imshow("Log Transformation", log_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

2. WAP to enhance the input image using power law/Gamma transform. Assume c=1, and input the value of gamma from the user.

```python
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg")
gamma = float(input("Enter gamma value: "))
img_normalized = img / 255.0
gamma_img = np.power(img_normalized, gamma) # Apply Gamma Transformation
gamma_img = gamma_img * 255
gamma_img = np.uint8(gamma_img)
cv2.imshow("Original Image", img)
cv2.imshow("Gamma Transformed Image", gamma_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

WAP a progra to enhance the input image using intensity levelslicing (constant stretchung).the inputrange of intensity values and desired range of intensity value in the output image are obtained from the user (example : the input range of intensity value from 80 to 120 is streched to 50 - 150 in the output image.

```python
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg")
r_min = int(input("Enter input minimum intensity (r_min): "))
r_max = int(input("Enter input maximum intensity (r_max): "))
s_min = int(input("Enter output minimum intensity (s_min): "))
s_max = int(input("Enter output maximum intensity (s_max): "))
img_float = img.astype(np.float32)
stretched = s_min + ((s_max - s_min) / (r_max - r_min)) * (img_float - r_min)
stretched = np.clip(stretched, 0, 255)
stretched = np.uint8(stretched)
cv2.imshow("Original Image", img)
cv2.imshow("Intensity Level Stretched Image", stretched)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
