Write a program to smooth the input image using:  
1. Averaging filter (un-weighted)
2. Weighted filter given by h(x,y)= max(|x|,|y|)
3. Gaussian filter
Note: Input the filter size and other parameters, such as sigma in case of Gaussian filter, method of padding (replication, zero padding, etc.) from the user.  

```python
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg")
if img is None:
    print("Image not found!")
    exit()
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print("Choose filter:")
print("1. Averaging Filter")
print("2. Weighted Filter")
print("3. Gaussian Filter")

choice = int(input("Enter your choice: "))

ksize = int(input("Enter filter size (odd number, e.g. 3, 5, 7): "))

if ksize % 2 == 0:
    print("Filter size must be odd!")
    exit()
print("\nChoose padding method:")
print("1. Zero Padding")
print("2. Replication")
print("3. Reflection")

padding_choice = int(input("Enter padding choice: "))
if padding_choice == 1:
    border_type = cv2.BORDER_CONSTANT
elif padding_choice == 2:
    border_type = cv2.BORDER_REPLICATE
elif padding_choice == 3:
    border_type = cv2.BORDER_REFLECT
else:
    print("Invalid padding choice!")
    exit()
if choice == 1:
    kernel = np.ones((ksize, ksize), np.float32)
    kernel = kernel / (ksize * ksize)
    result = cv2.filter2D(
        img,
        -1,
        kernel,
        borderType=border_type
    )
    title = "Averaging Filter"
elif choice == 2:
    center = ksize // 2
    kernel = np.zeros((ksize, ksize), dtype=np.float32)
    for i in range(ksize):
        for j in range(ksize):
            x = i - center
            y = j - center
            kernel[i, j] = max(abs(x), abs(y))
    kernel_sum = np.sum(kernel)
    if kernel_sum != 0:
        kernel = kernel / kernel_sum

    result = cv2.filter2D(
        img,
        -1,
        kernel,
        borderType=border_type
    )
    title = "Weighted Filter"
elif choice == 3:
    sigma = float(input("Enter sigma value: "))
    result = cv2.GaussianBlur(
        img,
        (ksize, ksize),
        sigmaX=sigma,
        borderType=border_type
    )
    title = "Gaussian Filter"
else:
    print("Invalid filter choice!")
    exit()
cv2.imshow("Original Image", img)
cv2.imshow(title, result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
