Problem 1 : WAP to draw a border around the input image (or in other words pad the input image).
1. Input the width of border(in terms of pixel count) and the desired color of the border from the user.
2. Try this for both binary and grayscale image.

For a grayscale image image

```python
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg", cv2.IMREAD_GRAYSCALE)
width = int(input("Enter border width (in pixels): "))
color = int(input("Enter border color (0-255): "))
rows, cols = img.shape
padded_img = np.full(
    (rows + 2 * width, cols + 2 * width),
    color,
    dtype=np.uint8)
padded_img[
    width:width + rows,
    width:width + cols
] = img
cv2.imshow("Original Image", img)
cv2.imshow("Image with Border", padded_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

For a binary image

```python
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg", cv2.IMREAD_GRAYSCALE)
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
width = int(input("Enter border width: "))
color = int(input("Enter border color (0 or 255): "))
rows, cols = binary.shape
padded = np.full(
    (rows + 2 * width, cols + 2 * width),
    color,
    dtype=np.uint8)
padded[width:width + rows, width:width + cols] = binary
cv2.imshow("Binary Image", binary)
cv2.imshow("Binary Image with Border", padded)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Problem 2 : WAP for image compliment Try this for both binary and grayscale image.

Binary Image Complement

```python
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg", cv2.IMREAD_GRAYSCALE)
_, binary = cv2.threshold(img, 127, 1, cv2.THRESH_BINARY)
L = 2
complement = L - 1 - binary
binary_display = binary * 255
complement_display = complement * 255
cv2.imshow("Binary Image", binary_display)
cv2.imshow("Binary Complement", complement_display)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Grayscale Image Complement

```python
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg", cv2.IMREAD_GRAYSCALE)
max_pixel = np.max(img)
bits = int(np.ceil(np.log2(max_pixel + 1)))
L = 2 ** bits
print("Maximum pixel value:", max_pixel)
print("Number of bits:", bits)
print("L:", L)
complement = L - 1 - img
cv2.imshow("Original Image", img)
cv2.imshow("Complement Image", complement)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
