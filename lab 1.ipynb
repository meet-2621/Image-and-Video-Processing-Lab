Read, display, and write an image.

```python
import cv2

img = cv2.imread("wallpaper.jpg")

if img is None:
    print("Error: Image not found!")
else:
    print("Image loaded successfully.")

    cv2.imshow("Original Image", img)
    cv2.imwrite("new_wallpaper.jpg", img)

    print("Image saved successfully.")
    print(img)   #(for Colored or RGB Images)
    cv2.waitKey(0)

    cv2.destroyAllWindows()
```

```python
import cv2
# Resize by Specifying Width and Height
img = cv2.imread("new_wallpaper.jpg")
resized = cv2.resize(img, (400, 300))
cv2.imshow("Original Image", img)
cv2.imshow("Resized Image", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Resize Using Scale Factors
img = cv2.imread("new_wallpaper.jpg")
resized = cv2.resize(img, None, fx=0.5, fy=0.5)  #fx = 0.5 → Width becomes 50% and fy = 0.5 → Height becomes 50%
cv2.imshow("Original", img)
cv2.imshow("Half Size", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Convert a colored image into a grayscale image.

```python
import cv2
img = cv2.imread("wallpaper.jpg")
if img is None:
    print("Error: Image not found!")
else:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    cv2.imshow("Original Image", img)
    cv2.imshow("Original Image", gray)
    cv2.imwrite("grayscale_wallpaper_image.jpg", gray)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

Read and display a video.

```python
import cv2
video = cv2.VideoCapture("Blood Supply.mp4")
if not video.isOpened():
    print("Error: Cannot open video!")
else:
    while True:
        ret, frame = video.read()
        if not ret:
            print("Video Finished")
            break
        cv2.imshow("Video", frame)
        if cv2.waitKey(25) & 0xFF == ord('q'):
            break
video.release()
cv2.destroyAllWindows()
```

```python
Extract frames from a video.
```

```python
import cv2
video = cv2.VideoCapture("Blood Supply.mp4")
count = 0
while True:

    ret, frame = video.read()
    if not ret:
        print("All frames extracted successfully!")
        break
    cv2.imwrite(f"frame_{count}.jpg", frame)
    print(f"Frame {count} saved.")
    count += 1
video.release()
cv2.destroyAllWindows()
```

```python
import cv2
video = cv2.VideoCapture("Blood Supply.mp4")
fps = video.get(cv2.CAP_PROP_FPS)
print("Video FPS:", fps)
frame_interval = int(fps / 2)
count = 0
saved = 0
while True:
    ret, frame = video.read()
    if not ret:
        break
    if count % frame_interval == 0:
        cv2.imwrite(f"frame_{saved}.jpg", frame)
        print(f"Saved frame {saved}")
        saved += 1
    count += 1
video.release()
cv2.destroyAllWindows()
print("Frame extraction completed.")
```
