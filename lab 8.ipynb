Write a program to generate following noises using equations derived
from PDFs of noise distributions. Compare your output with those
generated using inbuilt function imnoise. Plot the histograms of generated
noise to determine the shape of the distribution.
1. Uniform noise
2. Gaussian noise
3. Erlang noise
4. Exponential noise
5. Rayleigh noise

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Generate noise
size = (256, 256)

# 1. Uniform Noise
uniform = np.random.uniform(-30, 30, size)

# 2. Gaussian Noise
gaussian = np.random.normal(0, 20, size)

# 3. Erlang Noise
erlang = np.random.gamma(3, 5, size)

# 4. Exponential Noise
exponential = np.random.exponential(10, size)

# 5. Rayleigh Noise
rayleigh = np.random.rayleigh(10, size)

noises = [uniform, gaussian, erlang, exponential, rayleigh]

titles = ['Uniform', 'Gaussian', 'Erlang',
          'Exponential', 'Rayleigh']

# Display noise images and histograms
plt.figure(figsize=(12, 10))

for i in range(5):
    plt.subplot(5, 2, 2*i+1)
    plt.imshow(noises[i], cmap='gray')
    plt.title(titles[i] + ' Noise')
    plt.axis('off')

    plt.subplot(5, 2, 2*i+2)
    plt.hist(noises[i].ravel(), bins=50)
    plt.title(titles[i] + ' Histogram')

plt.tight_layout()
plt.show()
```

```python
# Built-in distribution functions

built_in = [
    np.random.uniform(-30, 30, size),
    np.random.normal(0, 20, size),
    np.random.gamma(3, 5, size),
    np.random.exponential(10, size),
    np.random.rayleigh(10, size)
]

plt.figure(figsize=(12, 8))

for i in range(5):
    plt.subplot(3, 2, i+1)
    plt.hist(built_in[i].ravel(), bins=50)
    plt.title('Built-in ' + titles[i])

plt.tight_layout()
plt.show()
```
