1. Write a program (WAP) to convert a grayscale image into a binary image.
Case 1: Assume the mean intensity of the image as the threshold value.
Case 2: Input the threshold value from the user.

```python
# Case 1
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg", cv2.IMREAD_GRAYSCALE)
if img is None:
    print("Error: Image not found!")
else:
    threshold = np.mean(img)
    print("Mean Threshold Value =", threshold)
    _, binary = cv2.threshold(img, threshold, 255, cv2.THRESH_BINARY)
    cv2.imshow("Grayscale Image", img)
    cv2.imshow("Binary Image", binary)
    cv2.imwrite("binary_mean.jpg", binary)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Case 2
import cv2
img = cv2.imread("wallpaper.jpg", cv2.IMREAD_GRAYSCALE)
if img is None:
    print("Error: Image not found!")
else:
    threshold = int(input("Enter Threshold Value (0-255): ")) # here user enters the threshold value
    _, binary = cv2.threshold(img, threshold, 255, cv2.THRESH_BINARY)
    cv2.imshow("Grayscale Image", img)
    cv2.imshow("Binary Image", binary)
    cv2.imwrite("binary_user.jpg", binary)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

2. Write a program (WAP) to convert an RGB (color) image into a grayscale image.
Case 1: Convert the RGB image to grayscale by taking the mean average of the three color planes (R, G, and B).
Case 2: Input the weightages of the three color planes (R, G, and B) from the user. The weightage of each plane should be a value between 0 and 1, and the sum of all three weightages must be equal to 1.
Example:
Weightages of R, G, and B are 0.7, 0.2, and 0.1, respectively.

```python
# Case 1
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg")
if img is None:
    print("Error: Image not found!")
else:
    B, G, R = cv2.split(img)
    gray = ((R.astype(np.float32) +
             G.astype(np.float32) +
             B.astype(np.float32)) / 3).astype(np.uint8) # Mean average method
    cv2.imshow("Original Image", img)
    cv2.imshow("Grayscale (Mean Method)", gray)
    cv2.imwrite("gray_mean.jpg", gray)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Case 2
import cv2
import numpy as np
img = cv2.imread("wallpaper.jpg")
if img is None:
    print("Error: Image not found!")
else:
    wR = float(input("Enter weight for R (0-1): "))
    wG = float(input("Enter weight for G (0-1): "))
    wB = float(input("Enter weight for B (0-1): "))
    if (0 <= wR <= 1 and 0 <= wG <= 1 and 0 <= wB <= 1 and
        abs((wR + wG + wB) - 1.0) < 1e-6):
        B, G, R = cv2.split(img)
        gray = (wR * R.astype(np.float32) +
                wG * G.astype(np.float32) +
                wB * B.astype(np.float32)).astype(np.uint8)
        cv2.imshow("Original Image", img)
        cv2.imshow("Grayscale (Weighted Method)", gray)
        cv2.imwrite("gray_weighted.jpg", gray)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("Invalid weights! Sum must be 1 and each value must be between 0 and 1.")
```
